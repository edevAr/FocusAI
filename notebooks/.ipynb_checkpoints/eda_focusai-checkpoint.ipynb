# FocusAI — Análisis Exploratorio de Datos (EDA)

Este notebook analiza el dataset de entradas de diario etiquetadas como **Productivo** o **Procrastinación**, y revisa los resultados del entrenamiento del modelo.

**Autor:** Luis Vasquez — Model Builder  
**Proyecto:** FocusAI MLOps Pipeline

```python
import json
import sys
from pathlib import Path

import matplotlib.pyplot as plt
import matplotlib.ticker as mticker
import numpy as np
import pandas as pd
import seaborn as sns
from collections import Counter

# Rutas
ROOT = Path("..")
sys.path.insert(0, str(ROOT))

RAW_CSV      = ROOT / "data" / "raw" / "journal_entries.csv"
CLEANED_CSV  = ROOT / "data" / "processed" / "cleaned_entries.csv"
FEATURES_CSV = ROOT / "data" / "processed" / "vectorized_features.csv"
METRICS_JSON = ROOT / "data" / "processed" / "metrics.json"
HOLDOUT_JSON = ROOT / "data" / "processed" / "holdout_metrics.json"
PERCLASS_JSON= ROOT / "data" / "processed" / "per_class_metrics.json"
COMPARISON_CSV = ROOT / "data" / "models" / "model_comparison.csv"
TUNING_CSV   = ROOT / "data" / "models" / "tuning_results.csv"

sns.set_theme(style="whitegrid", palette="Set2")
plt.rcParams["figure.dpi"] = 110
print("Librerías cargadas OK")
```

---
## 1. Vista general del dataset

```python
df = pd.read_csv(RAW_CSV)
print(f"Filas: {len(df)}  |  Columnas: {list(df.columns)}")
df.head(6)
```

```python
# Balance de clases
counts = df["etiqueta"].value_counts()

fig, ax = plt.subplots(figsize=(5, 4))
colors = ["#4CAF50", "#FF9800"]
bars = ax.bar(counts.index, counts.values, color=colors, edgecolor="white", width=0.5)
ax.bar_label(bars, fmt="%d", padding=4, fontsize=12, fontweight="bold")
ax.set_title("Balance de clases", fontsize=14, fontweight="bold")
ax.set_ylabel("Número de entradas")
ax.set_ylim(0, counts.max() + 8)
plt.tight_layout()
plt.show()
print(counts.to_string())
```

---
## 2. Análisis de longitud de texto

```python
df["n_chars"] = df["texto"].str.len()
df["n_words"] = df["texto"].str.split().str.len()

stats = df.groupby("etiqueta")[["n_chars", "n_words"]].agg(["mean", "min", "max"]).round(1)
print("Estadísticas de longitud por clase:")
display(stats)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

for ax, col, title in zip(axes, ["n_chars", "n_words"], ["Caracteres por entrada", "Palabras por entrada"]):
    for label, color in zip(["Productivo", "Procrastinación"], ["#4CAF50", "#FF9800"]):
        subset = df[df["etiqueta"] == label][col]
        ax.hist(subset, bins=12, alpha=0.7, label=label, color=color, edgecolor="white")
    ax.set_title(title, fontweight="bold")
    ax.set_xlabel(col.replace("n_", "").capitalize())
    ax.set_ylabel("Frecuencia")
    ax.legend()

plt.suptitle("Distribución de longitud por clase", fontsize=13, fontweight="bold", y=1.02)
plt.tight_layout()
plt.show()
```

---
## 3. Palabras más frecuentes por clase

```python
import os
os.environ["NLTK_DISABLE_IMPORT_SECURITY"] = "1"
import nltk
nltk.download("stopwords", quiet=True)
from nltk.corpus import stopwords

STOPWORDS_ES = set(stopwords.words("spanish"))
EXTRA_STOPS = {"si", "así", "tan", "ser", "hoy", "día", "todo", "toda", "nada", "vez"}
STOPWORDS_ES |= EXTRA_STOPS

def top_words(texts, n=15):
    words = []
    for text in texts:
        tokens = str(text).lower().split()
        words += [t.strip(".,;:¿?¡!") for t in tokens if t.strip(".,;:¿?¡!") not in STOPWORDS_ES and len(t) > 2]
    return Counter(words).most_common(n)

top_prod  = top_words(df[df["etiqueta"] == "Productivo"]["texto"])
top_proc  = top_words(df[df["etiqueta"] == "Procrastinación"]["texto"])

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
for ax, top, title, color in zip(
    axes,
    [top_prod, top_proc],
    ["Productivo — Top 15 palabras", "Procrastinación — Top 15 palabras"],
    ["#4CAF50", "#FF9800"]
):
    words, freqs = zip(*top)
    y_pos = range(len(words))
    ax.barh(y_pos, freqs, color=color, edgecolor="white")
    ax.set_yticks(y_pos)
    ax.set_yticklabels(words)
    ax.invert_yaxis()
    ax.set_title(title, fontweight="bold")
    ax.set_xlabel("Frecuencia")

plt.suptitle("Palabras más frecuentes por clase", fontsize=13, fontweight="bold", y=1.02)
plt.tight_layout()
plt.show()
```

---
## 4. WordCloud por clase

```python
try:
    from wordcloud import WordCloud

    fig, axes = plt.subplots(1, 2, figsize=(14, 5))
    for ax, label, color in zip(axes, ["Productivo", "Procrastinación"], ["Greens", "Oranges"]):
        text = " ".join(df[df["etiqueta"] == label]["texto"].tolist())
        wc = WordCloud(
            width=600, height=350,
            background_color="white",
            colormap=color,
            stopwords=STOPWORDS_ES,
            max_words=60,
        ).generate(text)
        ax.imshow(wc, interpolation="bilinear")
        ax.axis("off")
        ax.set_title(label, fontsize=14, fontweight="bold")

    plt.suptitle("WordCloud por clase", fontsize=13, fontweight="bold")
    plt.tight_layout()
    plt.show()
except ImportError:
    print("wordcloud no instalado. Ejecuta: pip install wordcloud")
```

---
## 5. Top features TF-IDF por clase

```python
features_df = pd.read_csv(FEATURES_CSV)
print(f"Shape features: {features_df.shape}")

feature_cols = [c for c in features_df.columns if c != "etiqueta"]

# Media TF-IDF por clase
mean_prod = features_df[features_df["etiqueta"] == 1][feature_cols].mean()
mean_proc = features_df[features_df["etiqueta"] == 0][feature_cols].mean()

# Diferencia normalizada para encontrar los más discriminativos
diff = (mean_prod - mean_proc).sort_values(ascending=False)
top_prod_feats  = diff.head(12)
top_proc_feats  = diff.tail(12)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
for ax, series, title, color in zip(
    axes,
    [top_prod_feats, top_proc_feats.sort_values()],
    ["Features TF-IDF → Productivo", "Features TF-IDF → Procrastinación"],
    ["#4CAF50", "#FF9800"]
):
    ax.barh(series.index, series.abs().values, color=color, edgecolor="white")
    ax.set_title(title, fontweight="bold")
    ax.set_xlabel("Diferencia media TF-IDF")
    ax.invert_yaxis()

plt.suptitle("Features más discriminativos (Productivo vs Procrastinación)", fontsize=13, fontweight="bold", y=1.02)
plt.tight_layout()
plt.show()
```

---
## 6. Comparación de modelos (CV)

```python
if COMPARISON_CSV.exists():
    comp = pd.read_csv(COMPARISON_CSV)
    display(comp.round(4))

    metric_cols = [c for c in ["Accuracy", "F1", "Precision", "Recall", "AUC"] if c in comp.columns]
    comp_melt = comp.melt(id_vars="Model", value_vars=metric_cols, var_name="Métrica", value_name="Score")

    fig, ax = plt.subplots(figsize=(10, 5))
    sns.barplot(data=comp_melt, x="Model", y="Score", hue="Métrica", ax=ax)
    ax.set_ylim(0, 1.05)
    ax.set_title("Comparación de modelos — Validación cruzada", fontsize=13, fontweight="bold")
    ax.set_xlabel("")
    ax.legend(bbox_to_anchor=(1.01, 1), loc="upper left")
    plt.tight_layout()
    plt.show()
else:
    print("model_comparison.csv no encontrado. Ejecuta el pipeline primero.")
```

---
## 7. Métricas hold-out (test set)

```python
if HOLDOUT_JSON.exists():
    holdout = json.loads(HOLDOUT_JSON.read_text(encoding="utf-8"))
    metric_keys = ["Accuracy", "F1", "Precision", "Recall", "AUC"]
    hm = {k: holdout[k] for k in metric_keys if k in holdout}

    fig, ax = plt.subplots(figsize=(7, 4))
    bars = ax.bar(hm.keys(), hm.values(), color="#5C6BC0", edgecolor="white", width=0.5)
    ax.bar_label(bars, fmt="%.3f", padding=4, fontsize=11, fontweight="bold")
    ax.set_ylim(0, 1.15)
    ax.set_title(f"Métricas hold-out ({holdout.get('n_test_samples','?')} muestras, test_size={holdout.get('test_size',0.2)})",
                 fontsize=13, fontweight="bold")
    ax.set_ylabel("Score")
    plt.tight_layout()
    plt.show()

    print("\nDetalle hold-out:")
    for k, v in holdout.items():
        print(f"  {k}: {v}")
else:
    print("holdout_metrics.json no encontrado. Ejecuta el pipeline primero.")
```

---
## 8. Matriz de confusión y métricas por clase

```python
if PERCLASS_JSON.exists():
    pc = json.loads(PERCLASS_JSON.read_text(encoding="utf-8"))
    cm = np.array(pc["confusion_matrix"])
    labels = pc["label_names"]

    fig, axes = plt.subplots(1, 2, figsize=(13, 4))

    # Matriz de confusión
    sns.heatmap(cm, annot=True, fmt="d", cmap="Blues",
                xticklabels=labels, yticklabels=labels,
                ax=axes[0], linewidths=0.5)
    axes[0].set_title("Matriz de confusión (CV)", fontweight="bold")
    axes[0].set_ylabel("Real")
    axes[0].set_xlabel("Predicho")

    # Métricas por clase
    report = pc["classification_report"]
    class_metrics = []
    for label in labels:
        if label in report:
            row = report[label]
            class_metrics.append({"Clase": label, "Precision": row["precision"],
                                   "Recall": row["recall"], "F1": row["f1-score"]})
    cm_df = pd.DataFrame(class_metrics)
    cm_melt = cm_df.melt(id_vars="Clase", var_name="Métrica", value_name="Score")

    sns.barplot(data=cm_melt, x="Clase", y="Score", hue="Métrica", ax=axes[1])
    axes[1].set_ylim(0, 1.15)
    axes[1].set_title("Precision / Recall / F1 por clase (CV)", fontweight="bold")
    axes[1].set_xlabel("")
    axes[1].legend(bbox_to_anchor=(1.01, 1), loc="upper left")

    plt.tight_layout()
    plt.show()

    print("\nClassification Report:")
    report_df = pd.DataFrame(report).T.round(3)
    display(report_df)
else:
    print("per_class_metrics.json no encontrado. Ejecuta el pipeline primero.")
```

---
## 9. Resultados del tuneo de hiperparámetros

```python
if TUNING_CSV.exists():
    tuning = pd.read_csv(TUNING_CSV)
    print(f"Combinaciones probadas: {len(tuning)}")
    display(tuning.head(10).round(4))

    fig, ax = plt.subplots(figsize=(8, 4))
    ax.plot(tuning.index + 1, tuning["mean_test_score"], "o-", color="#5C6BC0", label="F1 medio CV")
    ax.fill_between(
        tuning.index + 1,
        tuning["mean_test_score"] - tuning["std_test_score"],
        tuning["mean_test_score"] + tuning["std_test_score"],
        alpha=0.2, color="#5C6BC0"
    )
    ax.set_title("Tuneo de hiperparámetros — F1 por combinación (ordenado por rank)", fontweight="bold")
    ax.set_xlabel("Rank")
    ax.set_ylabel("F1 weighted (CV)")
    ax.xaxis.set_major_locator(mticker.MaxNLocator(integer=True))
    ax.legend()
    plt.tight_layout()
    plt.show()

    if METRICS_JSON.exists():
        metrics = json.loads(METRICS_JSON.read_text(encoding="utf-8"))
        best_params = metrics.get("best_hyperparams", {})
        if best_params:
            print("\nMejores hiperparámetros encontrados:")
            for k, v in best_params.items():
                print(f"  {k}: {v}")
else:
    print("tuning_results.csv no encontrado. Ejecuta el pipeline primero.")
```
